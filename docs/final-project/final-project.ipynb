# Final project: Aviapolis data kit and workflow

## Overview and purpose

In the practicals you have followed instructions step by step. In the final project you will plan and carry out the work yourselves.

You will work in groups on **Aviapolis**, the area of Vantaa that the Urban Challenge Studio (USP-E0501) will focus on in spring. Each group takes one district of Aviapolis and does two things:

1. **Build a data kit**: find, check, and organise the spatial data available for your district, so that you (and your classmates) can start the studio with data that is ready to use.
2. **Design and test a workflow**: draw a plan for answering an urban question about your district, carry it out in QGIS, and show how your results change when you change the unit or the area of analysis.

The project is not a repeat of the studio tasks. The aim is that you arrive in the studio already knowing where Vantaa's data is, how to get it into QGIS, and how to plan an analysis before you start clicking.

## The study area

Aviapolis is one of Vantaa's seven major districts (*suuralue*). It includes Helsinki Airport and the areas around it, and it is changing fast: business areas are turning into mixed-use neighbourhoods, and the Vantaa tram line will run through it.

Aviapolis consists of six districts (*kaupunginosat*). Each group is assigned one of them:

| District (fi) | District (sv) |

| Ylästö | Övitsböle |
| Pakkala | Backas |
| Lentokenttä | Flygfältet |
| Viinikkala | Vinikby |
| Veromies | Skattmans |
| Tammisto | Rosendal |

If there are more groups than districts, two groups may work on the same district. Group and district assignments will be announced on Discord.

You can explore the area in the City of Vantaa map service at [kartta.vantaa.fi](https://kartta.vantaa.fi/) before you start.

## Part 1: Data kit

Your data kit is a package that someone else could open and start working with straight away.

### What the data kit contains

1. **A QGIS project** with at least **8 layers** for your district:
   - at least **4 layers must come directly from the City of Vantaa's interfaces** (WFS or WMS, see below);
   - the other layers can come from other sources (Statistics Finland, HSY, HSL, OpenStreetMap, etc.);
   - the layers should cover at least **4 of these themes**: people, buildings and land use, services, green areas and nature, transport, planning and new development;
   - vector layers clipped to your district (or a sensible area around it) and saved in one GeoPackage;
   - all layers in the same coordinate reference system (we recommend **EPSG:3067, ETRS-TM35FIN**).
2. **A data table** describing each layer (see the template below).
3. **A one-page data guide**: short, step-by-step instructions that another student could follow to get the same data from the original source.

### City of Vantaa interfaces

The City of Vantaa publishes much of its spatial data through open interfaces. No login is needed for the open layers.

| Interface | Address to add in QGIS |
|---|---|
| WFS (vector features) | `https://gis.vantaa.fi/geoserver/wfs` |
| WMS (map images) | `https://gis.vantaa.fi/geoserver/wms` |
| WMTS (map tiles) | `https://gis.vantaa.fi/geoserver/gwc/service/wmts` |

More information: [gis.vantaa.fi/rajapinnat](https://gis.vantaa.fi/rajapinnat/).

To add the WFS in QGIS: **Layer → Add Layer → Add WFS / OGC API – Features Layer → New**, give the connection a name (e.g. *Vantaa WFS*), paste the address above, click **OK**, then **Connect**. You will see the list of available layers. The WMS works the same way through **Add WMS/WMTS Layer**. You practised this in the bootcamp.

Some WFS layers that are useful for Aviapolis:

| Layer | What it contains |
|---|---|
| `indeksit:kaupunginosat` | District boundaries (the field `suuralue` tells you the major district) |
| `indeksit:postinumeroalueet` | Postal code areas |
| `gis:rakennukset` | Buildings, with e.g. use (`kayttotarkoitus`), floor area (`kerrosala`), number of floors, number of dwellings (`huoneistojen_lkm`), and completion date |
| `taustakartta:maanpeiteaineisto_2018` | Land cover 2018 |
| `kaava:asemakaavat_vireilla` | Local detailed plans in preparation |
| `kaava:yleiskaavat_vireilla` | Master plans in preparation |
| `liikenne:ratikkalinja` | Vantaa tram line |
| `liikenne:liikenne_pyoratiet_tavoiteverkko` | Target cycling network |
| `gis:luo_matti_ekologinen_yhteys` | Ecological connections |

This list is a starting point, not a requirement. Look through the full list yourselves. The WMS also includes the master plan, historical aerial photographs (1954–2021), and old maps.

## Part 2: Workflow

### 1. Choose a question

Choose **one question** about your district that needs at least **three datasets** from your kit. Good questions are about a real planning issue and can be answered, at least partly, with your data. For example:

- How dense and how mixed is the built environment around the planned tram line?
- Where in the district is new construction happening, and what is it replacing?
- How much of the district is actually built up, and how does that change our idea of its density?
- Are residential buildings close to green areas and ecological connections?

You can use one of these or write your own.

### 2. Draw the workflow (Workshop I, 2 December)

In class, draw your workflow **by hand** on a large sheet of paper. Show:

- your question;
- the datasets you will use;
- the GIS steps, in order;
- the maps or figures you expect to produce;
- the **decision points**: where you have to choose a unit (density *of what*?) and a container (density *within what*?).

Other groups will walk around and leave comments on your sheet. Take a photo of the sheet with the comments. This is **version 1**.

### 3. Carry it out

Run your workflow in QGIS. Then change **either the unit or the container** and run it again. For example:

- **Unit:** number of buildings → floor area → number of dwellings
- **Container:** whole district → built-up area only (using the land cover layer) → 250 m grid → postal code areas

Your analysis must include at least one **density** measure (count or amount per area) or one **diversity** measure. A simple diversity measure is the Shannon index:

$$H = -\sum_{i=1}^{n} p_i \ln p_i$$

where $p_i$ is the share of category $i$ (for example, building use types) within the container. Higher values mean more variety.

### 4. Redraw the workflow

Once you are done, draw the workflow again by hand to show what you **actually** did. This is **version 2**. It is normal, and expected, that it differs from version 1.

### What you hand in for Part 2

- Photos of workflow **version 1** (with comments) and **version 2**;
- **at least two maps**: one with your first analysis and one after changing the unit or container;
- a short text (about **800–1,000 words**) that covers:
  - your question and why it matters for the district;
  - what you did and why you chose those data and methods;
  - what the two maps show and how the results changed when you changed the unit or container;
  - what changed between workflow version 1 and version 2, and why;
  - the limitations of your data and analysis.

## Part 3: Presentations and peer grading (Workshop II, 9 December)

All groups present on 9 December. **Everyone in the group must take part in the presentation.**

- **10 minutes** of presentation, followed by **5 minutes** of questions.
- Show your workflow drawings (version 1 and version 2), your data kit, and your maps.
- Be ready to answer questions about any step. Course staff may ask you to show where a number on your slide comes from in your QGIS project.

### Peer grading

You will grade the presentations of the **other groups**. Each student fills in a short grading form for each presentation, using the criteria below (0–5 points each). Your group's presentation grade is the average of the scores given by the other students. Course staff check the scores and may adjust clearly unjustified ones.

| Criterion | 0–5 points |
|---|---|
| **Question:** Is the question clear and relevant for the district? | |
| **Data:** Is the data kit well chosen and well documented? | |
| **Method:** Is the workflow logical, and is the change of unit or container explained? | |
| **Maps:** Are the maps clear and easy to read? | |
| **Reflection:** Does the group explain what changed and what the limitations are? | |

Each form also has space for one thing the group did well and one suggestion. Write these as you would like to receive feedback yourself.

Grading the other groups is part of your own grade (see *Grading*). If you cannot attend on 9 December, contact the course staff in advance.

## Grading

The final project is worth **30%** of the course grade and is graded out of **100 points**.

| Part | Points | Graded by |
|---|---:|---|
| Data kit (QGIS project, data table, data guide) | 30 | Course staff |
| Workflow, maps, and text | 30 | Course staff |
| Presentation | 20 | Other students (peer grading) |
| Peer grading of other groups (completed and thoughtful) | 5 | Course staff |

The group parts give the same points to all members. If a group member has clearly not contributed, the course staff may adjust that person's grade, based on the individual reflections and the presentation.

### What we look for

- **Data kit:** the layers work, are relevant, are documented honestly, and could be re-downloaded by someone else using your guide.
- **Workflow:** the question is clear, the steps follow from it, and you can explain your choices.
- **Maps:** they are well designed and support your argument (everything you have learned about map design applies).
- **Change of unit or container:** you show how the result changes and discuss what that means.
- **Reflection:** you are honest about limitations and about what changed from your plan.
