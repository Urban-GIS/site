# Final project: Planning a GIS analysis in Aviapolis

## Overview and purpose

In the practicals you followed instructions step by step. In the final project you will plan an analysis yourselves, from the question to the data, the methods, and the outputs.

You will work on **Aviapolis**, an area of Vantaa that is changing fast. You will choose a research question about one district, find and organise the data you need for it, and design a complete workflow for answering it.

You do **not** need to carry out the full analysis. Instead, you will show that your plan works by trying out a few parts of it. There is **no written report**: you present your work in a short presentation on **9 December**.

## Step 2: Choose a research question

Choose **one research question** about your district. The question can be about:

- **density and diversity**, two ideas that are central in urban planning (see the suggestions below);
- **any topic covered in the course**, such as green space, segregation, urban planning, accessibility, or air quality; or
- **something else that interests you**, as long as it can be studied with GIS.

It should be possible to answer your question, at least partly, with data that is available for your district. Browse [kartta.vantaa.fi](https://kartta.vantaa.fi/) and the layer list in Step 3 to get an idea of what exists.

Whatever your question, remember the two key choices from the course:

- **Unit:** *what* are you measuring? (people, dwellings, buildings, floor area, services, land cover types, trees...)
- **Container:** *within what* area? (the whole district, built-up land only, blocks, a 250 m grid, postal code areas, a distance around tram stops...)

Some suggested questions on **density and diversity**:

- Is the area around the planned tram line dense, diverse, both, or neither?
- Do the parts of the district with the most residents also have the most diverse building uses or services?
- How does the density of the district change if only built-up land is used as the container instead of the whole district?
- Are new buildings (e.g. completed after 2010) making the district denser, more diverse, or both?
- Where people live densely, is the surrounding green space also diverse (different land cover types), or mostly uniform?
- How does the picture of density and diversity change between postal code areas, districts, and a 250 m grid?

Some suggested questions on **other course topics**:

- How many residents live within walking distance of the planned tram line? (Practical 1, Practical 7)
- Do all parts of the district have access to green areas, and is anyone left out? (Practical 2, Practical 7)
- Are there sensitive sites, such as day care centres or schools, close to busy roads? (Practical 8)
- Where are new buildings and planned developments clustered? (Practical 5)
- How has the district changed since the 1950s? (historical aerial photographs in the Vantaa map service)

You can use one of these, adapt one, or write your own.

Your question can still change. If you find out in Step 3 that the data you need does not exist, or you find data that opens up a better question, refine your question and explain the change in your presentation.

## Step 3: Build a data catalogue

Now find the data you need to answer your question. A data catalogue tells you, and anyone else, what data you have, where it comes from, and what it can and cannot be used for.

### What the data catalogue contains

1. **A QGIS project** with at least **8 layers** that you need for your question (including, for example, boundaries and background maps):
   - at least **one layer must come from the City of Vantaa's map service**, [kartta.vantaa.fi](https://kartta.vantaa.fi/), loaded into QGIS through its WFS or WMS interface (see below). This is the City's own map service, so make sure you know how to work with it;
   - the other layers can come from other sources (Statistics Finland, HSY, HSL, OpenStreetMap, etc.);
   - vector layers clipped to your district (or a sensible area around it) and saved in one GeoPackage;
   - all layers in the same coordinate reference system (we recommend **EPSG:3067, ETRS-TM35FIN**).
2. **A catalogue table** with one row per layer (see the template below).

### City of Vantaa map service and interfaces

[kartta.vantaa.fi](https://kartta.vantaa.fi/) is the City of Vantaa's map service. You can browse it in your web browser, but to use its data in QGIS you need its open interfaces (WFS and WMS). No login is needed for the open layers.

| Interface | Address to add in QGIS |
|---|---|
| WFS (vector features) | `https://gis.vantaa.fi/geoserver/wfs` |
| WMS (map images) | `https://gis.vantaa.fi/geoserver/wms` |
| WMTS (map tiles) | `https://gis.vantaa.fi/geoserver/gwc/service/wmts` |

More information: [gis.vantaa.fi/rajapinnat](https://gis.vantaa.fi/rajapinnat/).

To add the WFS in QGIS: **Layer → Add Layer → Add WFS / OGC API – Features Layer → New**, give the connection a name (e.g. *Vantaa WFS*), paste the address above, click **OK**, then **Connect**. You will see the list of available layers. The WMS works the same way through **Add WMS/WMTS Layer**. You practised this in the bootcamp.

Some WFS layers that are useful for Aviapolis:

| Layer | What it contains |
|---|---|
| `indeksit:kaupunginosat` | District boundaries (the field `suuralue` tells you the major district) |
| `indeksit:postinumeroalueet` | Postal code areas |
| `gis:rakennukset` | Buildings, with e.g. use (`kayttotarkoitus`), floor area (`kerrosala`), number of floors, number of dwellings (`huoneistojen_lkm`), and completion date |
| `taustakartta:maanpeiteaineisto_2018` | Land cover 2018 |
| `kaava:asemakaavat_vireilla` | Local detailed plans in preparation |
| `kaava:yleiskaavat_vireilla` | Master plans in preparation |
| `liikenne:ratikkalinja` | Vantaa tram line |
| `liikenne:liikenne_pyoratiet_tavoiteverkko` | Target cycling network |
| `gis:luo_matti_ekologinen_yhteys` | Ecological connections |

This list is a starting point. Look through the full list yourselves, and choose what your question needs. The WMS also includes the master plan, historical aerial photographs (1954–2021), and old maps.

## Step 4: Plan the workflow

This is the main part of the project. Plan the **whole workflow** you would use to answer your question, from data to results. You do not need to carry it all out.

### Draw it first

Start by drawing the first version of your workflow **by hand** on paper, and take a photo of it.

Then improve the workflow. The final version can be drawn by hand again or made digitally (e.g. in PowerPoint, draw.io, or QGIS's Graphical Modeler).

### What the workflow must show

- **The question**, and the unit and container(s) you will use.
- **The data**: which layers from your catalogue you use in each step, and how they must be prepared (clipping, joining, filtering, reprojecting...).
- **The analysis steps**, in order, naming the tools you would use. Use the methods you learned during the course where they fit, for example:
  - buffers, clipping, and spatial joins (Bootcamp, Practical 1);
  - zonal statistics and field calculations (Practical 2);
  - efficiency ratios and density measures (Practical 3);
  - grid data (YKR) and choosing the right denominator (Practical 4);
  - hotspot analysis (Practical 5);
  - accessibility and network analysis (Practical 7);
  - automation with the Graphical Modeler (Practical 8).
- **How you measure what your question is about.** For example, density is a count or amount per area. For diversity, you can use the Shannon index, $H = -\sum_{i=1}^{n} p_i \ln p_i$, where $p_i$ is the share of category $i$ (e.g. building use types) within the container. Higher values mean more variety.
- **At least one alternative**: a point where you would change the unit or the container, and what you would compare.
- **The outputs** you would produce: which maps, charts, or tables, and what each one would show.
- **Risks and limitations**: what could go wrong, and what the data cannot tell you.

## Step 5: Try out parts of the workflow

You do not need to complete the analysis. Instead, choose **2–3 parts** of your workflow and carry them out, to show that your plan works (or to find out where it does not). For example:

- a map of one key measure for your district;
- the same measure with a different unit or container, side by side with the first;
- a model in the Graphical Modeler that does one part of the workflow;
- a test that shows a problem with the data, and how you would solve it.

If a test shows that part of your plan does not work, that is a useful result. Show it and explain how you changed the plan.

## Step 6: Presentation and peer grading (9 December)

Each group (or student working alone) presents for **max. 7 minutes**, followed by a few minutes of questions. If you work in a group, **everyone in the group must take part in the presentation**. Keep to the time: we will stop presentations at 7 minutes.

A suggested structure:

1. **Your question** (about 1 minute): your district and your research question, and why it interests you.
2. **Your data catalogue** (about 1.5 minutes): the layers you collected, where they come from (including the City of Vantaa map service), and their main limitations. What was missing, and did the data change your question?
3. **Your workflow** (about 2.5 minutes): your unit and container choices, and the workflow from data to outputs. Show how it changed from the first version.
4. **What you tried out** (about 2 minutes): the parts you carried out and what you learned from them.

Be ready to answer questions about any step. 

### Peer grading

You will grade the presentations of the **other groups**. Each student fills in a short form for each presentation, scoring each criterion below from 0 to 20, for a total of **0–100 points**.

| Criterion | 0–20 points |
|---|---|
| **Question:** Is the question clear, interesting, and suitable for a GIS analysis of the district? | |
| **Data catalogue:** Is the data suitable for the question and honestly described? | |
| **Workflow:** Is the workflow logical and complete, and are the unit and container choices explained? | |
| **Tryouts:** Do the parts that were carried out show that the plan works, or show clearly where it needs to change? | |
| **Presentation:** Is the presentation clear and well organised, and does it stay within 7 minutes? | |
| **Total** | **100** |

Each form also has space for one thing the group did well and one suggestion. Write these as you would like to receive feedback yourself.

### What we look for

- **Question:** the question is clear, relevant to the district, and possible to answer with available data.
- **Data catalogue:** the layers work, are what your question needs, and are documented clearly and honestly.
- **Workflow plan:** the steps follow from the question, the methods are suitable, and you can explain your choices of unit and container.
- **Tryouts:** the parts you carried out are done carefully and tell you something about your plan. Maps follow what you have learned about good map design.
