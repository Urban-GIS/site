# Learning goals

Urban GIS and Visual Tools develops the practical and critical GIS skills needed to investigate urban questions and communicate spatial evidence clearly. The course combines guided QGIS work with increasingly independent analysis, interpretation, and visual communication.

After completing the course, you will be able to:

- define a clear, well-justified research question that can be addressed using GIS;
- identify, evaluate, and select suitable spatial data and GIS methods for an urban research or planning question;
- use current approaches to spatial-data handling, analysis, and map visualisation in QGIS;
- interpret results critically, recognising the limits of data, methods, and visual representations;
- communicate GIS results clearly through well-designed maps, other relevant visualisations, and concise written interpretation;
- apply the methods introduced in the course independently and professionally in later studies, including USP studios and a Master's thesis; and
- find and assess additional datasets, documentation, and methodological guidance when addressing a new GIS problem.

## How the course supports these goals

The first two weeks provide a fast QGIS starter, establishing the practical foundations needed for the course. The subsequent practicals use urban examples to develop these skills further. As the course progresses, you will make more of your own choices about data, methods, visualisation, and interpretation.

The goal is not only to follow a sequence of software steps. It is to develop sound judgement about what a spatial question requires, how it can be investigated, and how the resulting evidence should be communicated.
