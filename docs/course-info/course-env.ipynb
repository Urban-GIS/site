# Course environment

The practical work in this course is carried out in [QGIS](https://qgis.org/), a free and open-source geographic information system. QGIS will be used throughout the QGIS bootcamp and in the later practical assignments.

Please install QGIS on your own computer before the first course meeting. Bringing your own laptop is strongly recommended for the hands-on sessions.

## Install QGIS

Download QGIS from the [official QGIS download page](https://qgis.org/download/). Choose the **Long Term Release (LTR)** rather than the newest regular release. The LTR is the most stable option and is the version used for the course materials.

At the start of the course, this is QGIS **3.44 LTR**. A later maintenance update within the same 3.44 LTR series is fine. Do not use QGIS 4 for this course unless the course site later says otherwise.

### Windows

1. Open the [official download page](https://qgis.org/download/).
2. Under **Windows**, download the **Long Term Version for Windows** standalone installer.
3. Run the installer and accept the standard installation options.
4. Open QGIS after installation and check that it starts normally.

### macOS

1. Open the [official download page](https://qgis.org/download/).
2. Under **macOS**, download the **Long Term Version for macOS** installer.
3. Open the downloaded installer and move QGIS to your Applications folder if prompted.
4. Open QGIS once before the first class. macOS may ask you to confirm that you want to open the application.

### Linux

Use the installation instructions for your distribution on the [official QGIS download page](https://qgis.org/download/). Install the current QGIS LTR package where your distribution offers that choice.

## Before the first session

Before the QGIS bootcamp begins, please make sure that:

- QGIS opens successfully on your computer;
- you can save files locally and have enough free disk space for course data;
- you bring your laptop and charger to class.

No additional plugins need to be installed in advance. Any plugins, data, or configuration steps needed for a particular practical will be introduced in the relevant course material.

## Keeping your work organised

Create one folder for the course on your own computer and keep the data, QGIS projects, exports, and reports for each practical together. Do not move or rename files after adding them to a QGIS project unless you also update the project paths.

Course-specific data and instructions will be provided with each practical.

## If you have problems

Try installing QGIS before the first meeting, so that any installation problem can be addressed early. If QGIS does not start or you are unsure which installer to choose, bring your laptop to the first session and let the teaching team know.
