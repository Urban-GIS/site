# Grading rubrics --> weekly updates!

To make the grading process transparent, we publish the grading criteria for the weekly practicals here. These criteria are intended to give you a clear idea of what we look for when assessing your work.

The criteria will be updated weekly as new practicals are released, so please check this page regularly.

## Practical 1: Grading Criteria

Your report is graded out of 50 points.

| Criterion                             | Points | What we look for                                                                                                                       |
| ------------------------------------- | -----: | -------------------------------------------------------------------------------------------------------------------------------------- |
| **Report structure & clarity**        |     10 | A clear, well-organized 2–3 page report that answers the required questions and explains your methods and decisions.                   |
| **Map quality & cartography**         |   12.5 | A clear final map with appropriate symbology, labels, legend, scale bar, and north arrow.                                              |
| **Process documentation**             |     10 | A clear explanation of the QGIS tools and workflow you used, including important decisions and challenges.                             |
| **Reflection & critical thinking**    |    7.5 | Thoughtful discussion of your results, the usefulness and limitations of the analysis, and connections to the course concepts.         |
| **Completeness & technical accuracy** |     10 | Required digitizing, data acquisition, buffering, spatial joins, table joins, calculations, and visualization are completed correctly. |

### What makes a strong submission?

A strong report does more than show that you completed the steps. It demonstrates that you understand what you did, why you did it, and what the results mean.

We value:

* Correct and complete GIS analysis
* Clear and effective maps
* A well-explained workflow
* Thoughtful interpretation of results
* Discussion of limitations and possible errors
* Meaningful reflection on what you learned

Different approaches are acceptable when they achieve the task appropriately and are clearly explained.

**Total: 50 points**

## Practical 3: Grading Criteria

Your report is graded out of 50 points.

| Criterion                       | Points | What we look for                                                                                                                                                                                                                                                                              |
| ------------------------------- | -----: | --------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Report & structure**          |     10 | A clear, well-organized 2–3 page report, in addition to visualizations. Includes the required maps, process description, analysis, and reflection.                                                                                                                                            |
| **Analysis process**            |     15 | Correct use of the Field Calculator and attribute joins. Logical calculation of efficiency ratios at both required spatial scales: small areas and the 250 m grid. Demonstrates understanding of the spatial relationships and geometric predicates used in the analysis.                     |
| **Results & map visualization** |     15 | At least two maps showing efficiency ratios at different spatial scales. Maps include a legend, scale bar, and north arrow. Symbology follows the efficiency-ratio classification in the practical instructions and is clear and interpretable. Additional layers support the interpretation. |
| **Reflection & discussion**     |     10 | Thoughtful discussion of whether building efficiency ratio is a sufficient indicator of urbanity, what it may miss, and how different spatial scales reveal or hide patterns. Also reflects on what was learned and how it was applied.                                                       |

### What makes a strong submission?

A strong report demonstrates both technical understanding and the ability to interpret your results.

We value:

* Correct and complete analysis
* Clear and effective cartography
* Understanding of the QGIS workflow
* Thoughtful interpretation of the maps and results
* Comparison of results at different spatial scales
* Awareness of limitations and possible errors
* Meaningful reflection on the week's learning

Different approaches are acceptable when they achieve the task appropriately and are clearly explained.

**Total: 50 points**

### Optional Bonus

Up to **2 additional points** are available for going beyond the core requirements, for example by completing the optional block-level efficiency analysis or demonstrating exceptional visualization or analytical depth.
