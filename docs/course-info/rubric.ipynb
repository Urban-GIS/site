# Grading rubrics --> weekly updates!

To make the grading process transparent, we publish the grading criteria for the weekly practicals here. These criteria are intended to give you a clear idea of what we look for when assessing your work.

The criteria will be updated weekly as new practicals are released, so please check this page regularly.

## Practical 1: Grading Criteria

Your report is graded out of 50 points.

| Criterion                             | Points | What we look for                                                                                                                       |
| ------------------------------------- | -----: | -------------------------------------------------------------------------------------------------------------------------------------- |
| **Report structure & clarity**        |     10 | A clear, well-organized 2–3 page report that answers the required questions and explains your methods and decisions.                   |
| **Map quality & cartography**         |   12.5 | A clear final map with appropriate symbology, labels, legend, scale bar, and north arrow.                                              |
| **Process documentation**             |     10 | A clear explanation of the QGIS tools and workflow you used, including important decisions and challenges.                             |
| **Reflection & critical thinking**    |    7.5 | Thoughtful discussion of your results, the usefulness and limitations of the analysis, and connections to the course concepts.         |
| **Completeness & technical accuracy** |     10 | Required digitizing, data acquisition, buffering, spatial joins, table joins, calculations, and visualization are completed correctly. |

### What makes a strong submission?

A strong report does more than show that you completed the steps. It demonstrates that you understand what you did, why you did it, and what the results mean.

We value:

* Correct and complete GIS analysis
* Clear and effective maps
* A well-explained workflow
* Thoughtful interpretation of results
* Discussion of limitations and possible errors
* Meaningful reflection on what you learned

Different approaches are acceptable when they achieve the task appropriately and are clearly explained.

**Total: 50 points**

| Criterion | Points | What we look for |

- Effective cartographic visualization of the results
- Correct interpretation of OD Matrix outputs
- Thoughtful discussion of individual travel behavior and nature preferences
- Meaningful reflection connecting the guest lecture to the practical

Different approaches are acceptable when they achieve the task appropriately and are clearly explained.

# Practical 9: Grading Criteria

*Prioritising Urban Green Areas: Composite Index for Central Helsinki*

Your report is graded out of 50 points.

| Criterion | Points | What we look for |
|---|---:|---|
| **Report & structure** | 10 | A clear, well-organized 2–3 page report that includes the prioritisation map, workflow description, data and indicators used, rankings, weighting, and discussion of the possibilities and limitations of composite indices. |
| **Indicator calculation & composite index** | 15 | Correct calculation and classification of green-area size, connectivity, and vegetation structure indicators. Appropriate spatial analysis and reclassification, followed by a correctly calculated weighted composite index and priority classification. Different but logically justified scoring, weighting, or classification schemes are acceptable. |
| **Map** | 10 | A clear prioritisation map of Central Helsinki's green areas, categorized into high, medium, and low priority. Includes a legend, scale bar, and north arrow, with clear and readable symbology. |
| **Interpretation & discussion** | 10 | Identifies and interprets the highest-priority green areas, discusses fragmentation and connectivity, provides reasoned recommendations for protection based on the actual results, and explains the chosen indicator rankings and weights. |
| **Reflection & method limitations** | 5 | Reflects on what was difficult, what was easy, and what was learned. Discusses the limitations and possibilities of Spatial Conservation Prioritisation (SCP) and composite-index methods, including assumptions, thresholds, weighting, and what the index may not capture. |

### What makes a strong submission?

A strong report demonstrates that you can combine multiple spatial indicators into a meaningful prioritisation framework and critically interpret the resulting patterns.

We value:

- Correct calculation of all three indicators
- Appropriate spatial processing and reclassification
- A logically constructed and justified composite index
- Clear and effective cartography
- Interpretation grounded in your own results
- Reasoned recommendations for prioritising green areas
- Critical discussion of the limitations and assumptions of composite-index approaches
- Meaningful reflection on the analytical process

Your interpretation should be specific to your results. Generic statements about spatial patterns, without referring to the actual index values, rankings, or mapped findings, are not sufficient for a strong discussion.

Different but logically sound indicator rankings, thresholds, weighting schemes, and classification approaches are acceptable when they are clearly justified.

**Total: 50 points**
