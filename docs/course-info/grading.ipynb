# Assessment and grading

Urban GIS and Visual Tools is graded on the University of Helsinki scale from **0 to 5**. Your grade reflects your ability to use GIS methods purposefully, communicate results clearly, and interpret urban questions critically.

## Assessment components

The final grade consists of:

- **Practical assignments (70%)**: weekly or thematic GIS exercises, normally submitted as short reports with maps and other relevant visualisations.
- **Final individual assignment (30%)**: an independent GIS-based analysis in which you select and apply appropriate data, methods, and visualisation techniques to an urban question.

Detailed requirements, submission dates, and any changes will be given on Moodle and in the relevant practical materials.

## What good work looks like

Assessment considers the appropriateness and care of your GIS work as well as how convincingly you communicate and reflect on the results. In practical reports and the final assignment, this includes:

- selecting and applying suitable data and GIS methods;
- presenting the main results clearly, using well-designed maps and other visualisations where relevant;
- explaining what the results show and acknowledging important limitations;
- writing a coherent report and documenting sources, data, and methods; and
- demonstrating increasing independence, critical thinking, and thoughtful use of the tools introduced during the course.

Work at the highest level goes beyond completing the technical steps. It makes well-justified choices, provides insightful interpretation, and communicates the analysis with clarity and care.

## Submission and late work

Submit each assignment by its stated deadline. If you expect to miss a deadline, contact the course assistants **before** the deadline whenever possible.

Unless the assignment instructions say otherwise, submit reports as a PDF. Keep your QGIS project and source files organised until the course has been completed, as you may need them to revise or clarify your work.

## Academic integrity and use of AI

All submitted work must represent your own learning and be prepared in accordance with the course guidance. The separate [AI tools page](ai-tools.ipynb) explains what kinds of AI support are permitted and how any use must be disclosed.
