# General information

Urban GIS and Visual Tools is a 5-credit course at the University of Helsinki, offered within the Urban Studies and Planning programme. The course combines GIS methods, visual communication, and urban questions through lectures and hands-on practical work in QGIS.

## Course website and communication

This open course website contains the learning materials, practical instructions, and course updates. Moodle is used for course administration and submissions.

Students should bring a laptop to the course meetings. The course uses **QGIS**, a free and open-source geographic information system. Installation and setup guidance will be provided on the course website.
