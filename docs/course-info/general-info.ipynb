# General information

Urban GIS and Visual Tools is a 5-credit course at the University of Helsinki, offered within the Urban Studies and Planning programme. The course combines GIS methods, visual communication, and urban questions through lectures and hands-on practical work in QGIS.

## Course meetings

The course runs from **2 September to 9 December 2026**. Meetings take place on **Wednesdays, 9:15–11:45**, across teaching periods 1 and 2. There are 14 meetings in total.

- **Usual location:** University Main Building, room **F3005**
- **2 December:** Language Centre, room **S115**
- **No meeting on 21 October**

Please check Sisu for the official implementation information and any changes to rooms or times.
