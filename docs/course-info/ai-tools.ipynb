# Use of AI tools


Artificial-intelligence tools based on large language models (AI-LLMs), such as ChatGPT, Google Gemini, and DeepL, can be useful when learning GIS. They can help explain concepts, troubleshoot QGIS workflows, and support data-processing tasks. However, their outputs may be inaccurate, incomplete, or unsuitable for the specific data, question, and spatial context at hand.

The purpose of this course is to develop your own ability to analyse urban spatial data and communicate the results clearly. You remain responsible for understanding, checking, and justifying every method, parameter, interpretation, map, chart, and written claim in your work.

## Guidelines for AI-LLM use in the course

    <strong>Note:</strong>
    <p>These guidelines apply to students completing the course for credit.</p>
</div>

You **can** use AI tools to support your learning, for example:

- to ask for explanations of GIS concepts, QGIS tools, or analytical methods;
- to help diagnose an error or think through a QGIS workflow;
- to get advice on organising, cleaning, or transforming data;
- to help you understand code or expressions used in QGIS, provided you test and understand them before use.

When you use an AI tool in a practical assignment, clearly state **how and why** you used it. Include this briefly in the report or other submitted work, for example: *"I used ChatGPT to help interpret an error in a QGIS expression. I checked and adapted the suggested expression before using it."*

You **cannot** use AI tools to produce material that you present as your own course work, including:

- written report text or interpretations;
- maps, charts, or other visualisations;
- completed analyses or answers to assignment questions.

Do not upload personal, confidential, or otherwise restricted data to public AI services. Do not treat a convincing AI response as evidence: check it against course materials, your data, and your own reasoning.

Failure to follow these guidelines may affect the grade for the relevant assignment.
