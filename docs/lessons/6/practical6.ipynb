## OVERVIEW & PURPOSE

Participatory sensing for GIS has become a useful tool to extract quantitative and qualitative information about the environment. Using map questionnaires is an easy way to collect the inhabitants’ experiential or other types of information about the everyday environment. Tools like Maptionnaire have enabled urban planners, researchers, and activist groups to gain insight about areas and phenomena. These tools have been used to find places/areas that feel, for example, safe, insecure, noisy, or quiet. Directing the map-based surveys towards a certain population group (children, the elderly, immigrants, etc.) can yield valuable data for planning and research purposes. Commonly, this way of obtaining spatial data through the public is known as **Public Participation GIS (PPGIS)**, **SoftGIS**, **Social/Participatory Sensing**. With the increasing presence of social media services and location services of smart technology, mining of social media data has become a frontier of “socially sensed” data.

**In this exercise, we explore data collected through a map-based survey conducted in 2024 on pleasant and unpleasant places and preferences for street greenery along active mobility routes in the Helsinki metropolitan area (HMA).** The survey was titled “Greentravel: A survey on your everyday active travel environment, and active mobility was defined for respondents in the survey as "non-motorised forms of travel such as walking or cycling that you use on a regular daily basis to get to a destination (e.g., go to work, school, shop, but excluding trips for leisure)".

The instructions in this exercise are less detailed, and you have more freedom to try out different options to solve the given GIS problem. **The problem to solve in this exercise is simple: look at the PPGIS data from the HMA and study people’s experiences.** Examine the distribution of points in relation to city districts, population distribution, and other potential variables. Try to find an explanation for the distribution of positive and negative experience points of different types. **The aim of this exercise is to be creative and try out different GIS possibilities.** So, don’t be afraid and try different methods. **However, you should do at least what is suggested in this exercise.**

### Data points and population in different areas

Population distribution, along with environmental variables, may affect the distribution of the data points. In this task, we want to know how many people live around the PPGIS survey place markings. We will use the population grid data and a spatial join. In spatial join, the cases are matched based on a spatial relationship, for example, an intersection.

**Note:** If you do not want to use population data for your analysis, feel free to use other relevant data.
    
8. Use the *Join by Attributes by Location* (Vector general) tool to combine the population information with the PPGIS survey point data:

    [picture]

    **Input layer**: The layer to which we want to join new data (data points)  
    **Join layer**: The layer that will be joined to our input layer (population grid)  
    **Geometric predicate**: The spatial relationship used as the criterion for joining (let’s use Intersects)  
    **Fields to add**: You can choose whether you want all fields from the join layer to be added to the input or only some. Here we only need the “vaesto” field. You can select that if you want to keep your attribute table less crowded.  
    **Join type**: This determines how the algorithm will act when an input feature matches multiple features in the join layer. Let’s go with one-to-one (although it will not make any difference in our case, as the grid cells do not overlap, and each point will fall into one cell only).  
    **Discard records which could not be joined**: This will remove features that could not be joined
    **Joined field prefix**: A prefix of your choice to be added to the new field names  
    **Joined layer**: The join output  
    **Unjoinable features from first layer**: If you want, you can produce a layer with all features that failed to join

9. Use the new layer and the newly added “vaesto” (population) field to make a map of the data points based on the areas’ population. Perhaps the obvious choice is a graduated color or symbol size map, but feel free to get creative.
