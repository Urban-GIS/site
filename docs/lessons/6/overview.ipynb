# Overview

## Learning objectives

**Lecture** 

**Practical**
1. Explore and prepare PPGIS data: download, inspect, and clean the GREENTRAVEL survey data, and understand its structure, attributes, and limitations (e.g. respondent profile, point vs. route data, place categories).
2. Combine participatory data with contextual data: integrate the PPGIS points with city districts, the population grid, and other open datasets through joins, spatial joins, and aggregation.
3. Analyse spatial patterns of experiences: compare the distributions of pleasant and unpleasant places and street greenery preferences using counts, densities, normalisation (e.g. points per resident), and simple statistics.
4. Interpret and explain the patterns: identify possible explanations for the observed distributions (e.g. land use, greenery, traffic, population density), while separating real patterns from sampling effects.
5. Reflect critically on PPGIS: assess whose perspectives the data represent, and what this means for using it in planning and research.
