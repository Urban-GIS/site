# Overview

## Learning objectives

**Lecture** 

- Explain the principles and applications of Public Participation GIS (PPGIS).
- Understand how participatory mapping data are collected, prepared, and analyzed.
- Recognize the opportunities and limitations of PPGIS data.
- Distinguish between place-based and person-based spatial analysis.
- Apply basic spatial statistics to identify patterns in PPGIS data.
- Explore and visualize participatory mapping data using QGIS.


**Practical**
- Explore and prepare PPGIS data: download, inspect, and clean the GREENTRAVEL survey data, and understand its structure, attributes, and limitations (e.g. respondent profile, point vs. route data, place categories).
- Combine participatory data with contextual data: integrate the PPGIS points with city districts, the population grid, and other open datasets through joins, spatial joins, and aggregation.
- Analyse spatial patterns of experiences: compare the distributions of pleasant and unpleasant places and street greenery preferences using counts, densities, normalisation (e.g. points per resident), and simple statistics.
- Interpret and explain the patterns: identify possible explanations for the observed distributions (e.g. land use, greenery, traffic, population density), while separating real patterns from sampling effects.
- Reflect critically on PPGIS: assess whose perspectives the data represent, and what this means for using it in planning and research.
