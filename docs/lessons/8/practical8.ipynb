## GIS automation and spatial analysis as a process

## OVERVIEW & PURPOSE

The ultimate goal in GIS is to learn how to solve problems spatially. **Spatial analysis is a process in which you examine geographical data, model problems geographically, and derive results by computer processing. Finally, you analyse the results.** Throughout this course, you have familiarized yourselves with multiple analysis methods and learned how to access different data sources, modify the data, and draw meaningful results. To be able to conduct your own spatial analyses in the future, it is important to answer the questions  
“What is the research question I want to answer?”, “What kind of data do I need?, "What is the input and output?”, and “What are the steps in between?”.

**Today, we are looking at spatial analysis as a process comprising data exploration, GIS-based modeling, and problem-solving.** We will get familiarized with QGIS’s *Graphical Modeler* -tool, which can be used to create models and tools of one's own. Models are visual workflows that connect input data with different geoprocessing tools and feed the output data of one tool into another tool as input, and produce, in the end, the desired output. This practice aims to understand the spatial analysis process as a **‘PPDAC’** process (Problem, Plan, Data, Analysis, Conclusion). That means seeing statistical research (such as GIS research) as a cycle, as the solution of one problem often leads to new questions.

### What are the ‘Processing Models’ in QGIS?

You can do a multi-step GIS analysis manually, or you can create a model, a chain of processes wrapped into a single process, which eliminates a good chunk of the work and saves a lot of time. Models also help you to repeat the same analysis for different data. Both QGIS and ArcGIS have model builders that allow the user to daisy-chain functions. There are many online tutorials that help you further understand automating GIS processes. Models can also be exported to a Python script if you’re interested in learning how to utilize the Python programming language in GIS.
