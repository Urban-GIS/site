## Assessing the quality of urban environments: Map questionnaire & web map

## OVERVIEW & PURPOSE

Assessing the quality of urban environments is a multi-faceted topic of major interest in urban planning, research, and design. Components and features of good urban environments have been fiercely debated throughout the 20th century, and the 21st doesn’t seem to buck the trend. Your task today is to evaluate public space near our classroom using the EpiCollect5 mobile phone app and a questionnaire. This might also be helpful if you’re venturing out to conduct field work during your studies. The results of your evaluation excursion will then be fashioned into an upload-ready web map using QGIS.

EpiCollect5 is a free tool for spatial data collection developed by Imperial College London and available on Android and Apple devices. It can be found from app stores by searching for EpiCollect5. You can also use it on your computer to create your own map questionnaires. You’ll have to sign up or use your Google account to create your own questionnaires.

Web maps are used daily by millions of people worldwide for various purposes. Public transport journey planners, nearest restaurant searches, and navigation tools nearly always use some sort of web map in the background. Creating web maps isn’t restricted to large companies or organizations, and thus, you too can create one and publish it online. In terms of open science (and activism), publishing interactive maps online is seldom used, and its full potential is still unrealized. A nice example of an interactive web map is Accessibility Research Group’s My Accessibility Planner: [www.mapple.fi](www.mapple.fi). Web maps are great tools for sharing important analysis results on a map and saving the layperson from downloading GIS software, learning to use it, and then being able to explore the data and the results of your analysis. Today, you’ll take the first step into the world of web maps.

## EXERCISE PHASES

### Gathering the data

1. Install Epicollect5 on your mobile phone.
    1. Download the application from the App Store (iPhone) or Google Play (Android).
    2. More info: https://five.epicollect.net/more-collect.
2. Pick at least 5 close-by spots to evaluate and head out to evaluate them using EpiCollect5.
    1. Try to stay within the area designated in the picture below.
    2. Zoom group participants may evaluate their home surroundings, etc.
3. Open EpiCollect5 and search for “USP303” to find the survey. This way, it’s ready to go when you arrive at your locations.
    1. GPS location most likely won’t work properly if you’re indoors. If you want to evaluate an area indoors, evaluate the spot first indoors. Then go outside for the final step of the evaluation, which is updating your GPS location.
4. Answer the survey at least at five locations.
    1. Make sure to read the questions properly before answering them.
    2. In the *Update location* phase, pay attention to the *accuracy* value. This is your GPS accuracy in meters. If this value is very high, push the update button again, perhaps several times. A few meters is very good, over 20 meters is bad.
    3. After answering and saving the answer, there are still a few “upload” buttons to be pressed before the data is submitted to the server.
5. Return when ready or at the latest by 10.00 to the classroom.

### Downloading and examining the data
6. The teachers will upload the results into Moodle when everyone has finished at around 10.00.
    1. Download the data (CSV file) onto your computer from Moodle or try directly from https://five.epicollect.net/project/usp3032022/data.
    2. Add the CSV file into QGIS using *Add delimited text layer*
        1. The coordinate system of the file is WGS84 (EPSG: **4326**)
        2. Use fields with beginnings “Long_” and “Lat_ “ (X is for longitude and Y is for latitude). There are also coordinate fields starting with UTM, but we don’t use them now.
        3. After successfully importing the CSV file, save it as a GeoPackage on your computer.
7. Open the attribute table of the file and explore it.
    1. The file is raw output from EpiCollect5.
8. Add a background map.
9. Do some visualisation experiments with this data. Try to visualise different questions asked in the form.
10. Use the legend fields when making the symbology. This defines the legend to be shown in the online map.

### Creating a web map

11. Create an account on QGIS Cloud.
    1. Go to https://qgiscloud.com/, and create a new account (*Log in* and *Sign up*).
    2. Check your email for a confirmation message and confirm your account.
12. Install the QGIS Cloud plugin and create a database.
    1. Log in to the plugin with your new account in the *Account* tab.

       [picture 2]

    2. Create a database by clicking *Create*. This may take a few minutes. (You can have only one database in the free account, but it is enough for the exercise.)

       [picture 3]

13. Upload data to the QGIS Cloud server.
    1. Switch to the *Upload Data* tab and click *Upload Data*. The local map layer on your project is now uploaded to the server. You will be asked for a name for your modified project. (Use only the point layer vector file. Rasters won’t fit into the space available in the free account.)
    2. Add a background layer for your map for the QGIS cloud plugin. The ones in the QuickMapServices may work, but it is best to use one of the maps in the QGIS Cloud background map menu. Remove any background map you used earlier.
14. Publish the web map.
    1. Go to the *Maps* tab and click *Publish Map*. You may be asked to change your project coordinate system to EPSG 3857, which will work best on web maps.

       [picture 4]
                                                                                                                                                                                            
    2. You can acquire and open the link to your web map from the QGIS Cloud webpage.

       [picture 5]

    3. If you need to change your map after publishing, it is best to delete the map from published maps, click *Refresh* in the account tab, and publish the map again.
