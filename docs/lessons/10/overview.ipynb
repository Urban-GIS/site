# Overview

## Learning objectives

- Collect spatial data in the field with a mobile map questionnaire (EpiCollect5)
- Evaluate the quality of public space in a systematic way, using a shared set of questions
- Recognise how GPS accuracy affects the location of data points, and check accuracy while collecting data
- Import questionnaire results (CSV) into QGIS using coordinate fields, and save them as a GeoPackage
- Explore and visualise different questionnaire variables, and design symbology and legends for an online map
- Publish an interactive web map with QGIS Cloud and share it through a link
- Reflect on the strengths and limitations of field-collected, GPS-based data
