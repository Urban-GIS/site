# Practical 1

## Digitization and proximity analysis of RaideJokeri light rail stops

### OVERVIEW & PURPOSE

In this exercise, you will first create your own vector data by digitizing different features. Then you will perform proximity analysis to estimate how many people live within walking distance of the new RaideJokeri light rail stops. This will involve combining open data from different sources, creating walking-distance buffers around the stops, and using spatial joins to calculate the total population within each buffer zone. This is a common and practical type of GIS analysis used in urban planning and transportation studies.  
By doing this exercise, you will:  
- Practice digitizing points, lines, and polygons and saving them as vector data  
- Learn to acquire and combine data from multiple open data servers (WFS)  
- Master the process of creating buffers and performing spatial joins to aggregate data  
- Understand how to use table joins to effectively visualize results on a map  


### OBJECTIVES  
- Digitize different urban features near selected light rail stops  
- Acquire spatial datasets from geospatial servers  
- Create buffers representing fixed walking distances around light rail stops  
- Join population data to buffer zones using spatial and table joins

- Answer guiding questions to reflect on your process and learning

**Creating walking distance buffers**


**12.** Next up, we will create buffers around each stop. Buffering tools are found in the
Processing toolbox by typing buffer into the search bar or from Vector > Geoprocessing
Tools > Buffer.

**13.** Set RaideJokeri_pysakit layer as the Input layer and distance to what you deem to be a
reasonable walking (straight line) distance to a public transportation stop from home –
for example, 400 meters. Set segments to 20. Don’t check the dissolve option. Then click
Run and a layer called Buffered should be added to the project. You have now created
zones depicting walking distance around all of the RaideJokeri stops. Make sure to make
the layer permanent (Properties>Make Permanent).

**14.** Check out the Attribute table of Buffered and compare it with the RaideJokeri_pysakit
layer. You can see that the attribute table is identical: the only difference is that now the
vector features are polygons of set size instead of points. This will come in handy later.


**Performing a spatial join with population data**

**15.** **CRITICAL STEP:** Ensure all your layers (stops, buffers, population grid) are in the same
Coordinate Reference System (CRS), preferably ETRS89-TM35FIN (EPSG:3067). Use
Export > Save Features As... to reproject any layers that are not.

**16.** In the Processing Toolbox, search for and open *Join attributes by location (summary)*.

- Set the Buffered layer as the Base layer.
- Set the *Väestotietoruudukko_2024* layer as the Join layer.
- Set the Geometric predicate to *intersect*. This field contains the summary of
population in that cell.
- In Fields to summarise, select *asukkaita* (the population count field). For
Summaries, **choose sum.**
- Check *Discard records* which could not be joined.
- Run the algorithm. A new Joined layer is created with a new *asukkaita_sum* field
containing the total population within each buffer.

<img src="figures/buffer400.png" width="500"> 

**17.** A new layer called Joined layer should now have been added to the project. Visually the
new layer looks similar to the buffer layer, but check the attribute table and you will find
a new field, asukkaita_sum, which depicts the total population count within each buffer.
Make the layer permanent. Great! We have the data we need.

**Table join**

**18.** To visualize the population data on the original stop points, perform a table join. Use the
*Join attributes by field value* tool from the Processing Toolbox.
- Set Raidelokeri_pysakit as Input layer and the Joined layer from the previous step
as Input layer 2.
-  For both layers, set the Table field to id (the unique identifier for each stop).
- In Layer 2 fields to copy, select asukkaita_sum.
- Save the output as a new shapefile (e.g., stops_with_population.shp).

<img src="figures/fieldvalue.png" width="500"> 



**Visualization**

**19.** We want to visually show ***how populated the proximity of each stop is.*** Style the new
stops_with_population layer to show the population count.

- Open the layer's Properties > Symbology.
- Change the symbology type to Graduated.
- Set the Value to asukkaita_sum.
- Change the Method to Size to represent larger populations with larger symbols.
- Click Classify and then Apply. Adjust the sizes and colors to create a clear and
effective visual.
