# Overview

## Learning Objectives

**Lecture**
- Classify data for effective mapping (equal interval, quantiles, natural breaks)
- Apply the geographic inquiry process to solve GIS problems
- Recognize core GIS tools: spatial join, clipping, digitizing, buffering

**Practical**
- Digitize points, lines, and polygons
- Pull vector data from WMS/WFS servers
- Buffer features and spatially join population data
- Table-join results back to points
- Build a complete, styled map
