# Practical 7: Spatial Accessibility and GIS

## Various approaches to determine accessibility

### OVERVIEW & PURPOSE

Accessibility is one of the central research topics of urban geography. It characterises the physical and functional structure of a city as well as everyday choices available for individuals. Accessibility, often measured by travel time, is a useful concept both in research and planning practice because it links questions related to land use and transport. It has also led to the concept of the **15-minute city** urban planning model, where everything a resident needs in their daily life can be accessed within a 15-minute walk or bike ride.

However, accessibility also depends on people and their individual preferences based on abilities, needs and lifestyle choices. This allows us to look into individual needs, which vary across urban populations as cities are becoming more diverse. In this practical, we will look into accessibility modelling using the components of travel time as well as individual lifestyles.

### Task 2 (Optional): Travel Time Matrix and the 15-minute city

> **Optional:** This task is not required, but completing it has a positive effect on your grade.

Helsinki Region Travel Time Matrix is an open spatial dataset produced by the Digital Geography Lab at the University of Helsinki. The dataset contains travel time and distance information for routes between all 250 m x 250 m YKR grid cell centroids (n = 13 132) in the Helsinki Metropolitan Area by walking, cycling, public transport and car.

***Determining YKR-ID***

*The Travel Time Matrix is created using centroids of YKR grid cells (Yhdyskuntarakenteen seurannan tietojärjestelmä); this makes it possible to combine accessibility information with a range of socio-economic variables (concerning population, jobs, etc.) stored at the same spatial resolution in the YKR data.*

One file in the matrix describes travel times and travel distances from all YKR grid cells (13 230 cells) to one particular cell. For instance, the “..travel_times_to_5975373.csv” file includes travel times and distances from all other cells to the cell number 5975373 (the number is based on ID codes of YKR data). First, we need to find out in which YKR grid cell our shopping centre, Itis, is located.
