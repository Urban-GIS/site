# Overview

## Learning objectives

**Lecture** (Guest lecture by Arpa Aishwarya - University of Turku)
- Define spatial accessibility and explain why it matters for planning, housing and business location, and social and environmental equity
- Explain the 15-minute city concept and its role in public and planning debates
- Identify the four components of accessibility: land use, transport, time and people (Geurs & van Wee 2004)
- Compare measures of travel cost: Euclidean vs. network distance, calculated vs. perceived travel time, and other costs
- Describe the Helsinki Region Travel Time Matrix: travel modes, time of day, person-sensitive speeds and its YKR grid structure
- Explain person-based accessibility through activity space models and why availability does not mean accessibility
- Explore how PPGIS data reveal differences in access to everyday nature, e.g. between income groups in Turku

**Practical**
- Import a road network from OpenStreetMap with QuickOSM
- Calculate Euclidean and network distances between home and nature points with QNEAT3
- Interpret individual accessibility using PPGIS data on everyday nature visits
- *(Optional)* Locate a destination in the YKR grid and join the Travel Time Matrix to the grid
- *(Optional)* Visualise travel times to a shopping centre by different travel modes
- *(Optional)* Estimate how many residents can reach the shopping centre within 15 minutes
