# Overview

## Learning Objectives

**Lecture**
- Explain color as physical light, human perception, and aesthetic experience
- Use hue, value, and saturation to build contrast and visual hierarchy
- Distinguish color schemes: analogous, complementary, polychrome, monochrome, neutral
- Apply map color considerations: contrast, blending, connotation, color blindness, restraint
- Define building efficiency ratios at plot, block, and regional scales
- Explain why grids are used in spatial analysis and how they relate to the MAUP
- Choose geometric predicates (intersects, contains, within) for spatial joins

**Practical**
- Extract features by expression and clean attribute data
- Calculate areas and efficiency ratios with the Field Calculator
- Create a 250 m grid with MMQGIS
- Summarize building floor area to the grid (join attributes by location)
- Classify and style efficiency ratios using TKK thresholds
- Critically assess whether efficiency ratios capture "urbanity"
- *(Optional)* Map block-level efficiency and build a KDE heatmap
