### OVERVIEW AND PURPOSE

What does "urban" really mean? Is it just tall buildings, or is it about the intensity of how we use

This is both a technical GIS task and a conceptual exercise: while you’ll practice calculations and

*Before moving on to the exercise, first let’s understand the concept of efficiency ratios.*
An efficiency ratio is a simple but powerful measure. it's the **total floor area of buildings** divided
by **the area of the land they sit on.**

$$
e = \frac{\text{total floor area}}{\text{total land area}}
$$

The piece of land can be the real estate, a neighborhood, a city district or any specified area A
high ratio indicates dense, "urban" development, while a low ratio suggests a more suburban
character. Conversely, **“sparsity ratio”** is the total area of all non-built features to the size of the
piece of land.

- *Regional efficiency* (ea): Measures how much built floor area exists compared to the total area of a neighborhood, district, or city. Shows how dense or urban the area is overall.  
- *Block efficiency* (ek): Shows how compact buildings are together (e.g. a block or cluster of plots)
- *Plot efficiency* (et): A single property

**Calculating efficiency on a 250-meter grid**

**8. Download the MMQGIS plugin through the Plugin manager.**  
*You should now be able to find a drop-down menu named MMQGIS at the top of the QGIS main window.*

**9. Create a grid using MMQGIS:**  
1. Create > Create Grid Layer.
2. Use the following specs:
   - Geometry type: Rectangles (others also ok)

   - Output file name: Helsinki_grid_250rec (for example)


*You can always check what the layer unit is in the Information tab of Layer properties. Because the unit here is meters and we want 250-meter by 250-meter grid cells, we can use ‘layer units’ as the ‘unit’ in the MMQGIS tool.*

**10. To prevent unnecessary crashes and the operation taking an eternity, export the building registry

**11. Join the building registry into the grid layer:**  
- Find the Join attributes by location (summary) tool from the processing toolbox.  
- Because you want to join building data into the grid, the grid will be your Base layer (or Input

- You want information based on where floor area spatially coincides with your grid elements – so
choose *intersects* or some other suitable geometric predicate.  
- Choose *KERALA2* as Fields to *summarize*.  
- Choose *sum* as *Summaries to calculate.*  
- Name the joined layer GridBuild250, for example.  
- Click *Run*.



**12. Create a new a field for the building efficiency ratios in the Field calculator of your new joined layer

- Output length: 10
- Precision: 2
- Expression: KERALA2 / (250 * 250)  
*We divide KERALA2 by 250 times 250 since this is the size (in meters) of each cell in our grid.*
