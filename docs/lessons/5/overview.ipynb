# Overview

## Learning objectives

**Lecture** (Guest lecture by Viva Amani Fard - City of Helsinki)
- Describe the Finnish urban planning system and its different planning levels
- Explain how GIS supports urban planning and spatial decision-making
- Identify GIS methods used for accessibility analysis, transport planning, and land-use analysis
- Explain how land-use simulation can support urban planning and future scenarios
- Distinguish artificial intelligence, machine learning, deep learning, and large language models
- Explore practical applications of GIS and AI in Helsinki's urban planning

**Practical**
- Explain the principles of hotspot analysis and Kernel Density Estimation (KDE)
- Create raster heatmaps from road traffic accident data in QGIS
- Examine how search radius, kernel shape, and weighting affect heatmap results
- Generate and compare accident hotspots for different road user groups
- Create a diagram showing accident involvement by road user and vehicle type
- Critically interpret spatial patterns of road traffic accidents in the Helsinki metropolitan area
