# Overview

## Learning objectives

**Lecture** (guest lecture by Mansi Garg)

- Understand YKR urban structure grid data: grid sizes, data groups, and field prefixes
- Calculate proportions with the right denominator and handle privacy-suppressed values (-1)
- Explore applications of grid data in accessibility, equity, and socio-economic indices
- Apply best practices for visualizing grid maps (subdued gridlines, sequential colors, map elements)

**Practical**

- Clean suppressed values with Select by Expression
- Calculate proportions with the Field Calculator
- Reclassify data into quartiles with conditional expressions
- Identify areas in the most disadvantaged quartile for both education and income
- Build a styled print-layout map
- Write a short report reflecting on segregation and data limitations
