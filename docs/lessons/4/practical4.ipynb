## The City and its People: Patterns of Segregation

### OVERVIEW & PURPOSE

Thus far we’ve learned about building efficiency ratios as indicators of urban intensity and next
week we will talk about clustering of businesses, but cities are more than that. One of the key
components of the city is missing and as Shakespeare so eloquently put in Coriolanus:

*“What is the city, but the people?”*

The socio-economic characteristics of a city's inhabitants vary across geographical space and
time. Understanding these patterns is key to analyzing topics like social segregation and equity.

In this exercise, you will examine the socio-economic structure of the Helsinki Metropolitan Region using postal code area data from Statistics Finland’s PAAVO 2025 dataset. Your task is to identify and map postal code areas with the lowest levels of education and income.  This involves data cleaning, calculating proportions, and
reclassifying data using conditional statements—skills that are fundamental to spatial analysis

This practical will help you move from raw data to meaningful spatial insight, encouraging you to
think critically about what these patterns reveal about the city and the limitations of the data
itself. The results are once again to be visualized on a map and to be analyzed in the report.

### EXERCISE PHASES

**Getting familiar with the data**

*Before working with any dataset, it's crucial to understand what information it contains and how
it's structured. This prevents errors and helps you plan your analysis effectively.*

**1.** Download, unzip and open the practical 4 data in QGIS

**2.** Explore the attribute table of the grid data. Use the provided PDF to understand the
Finnish field names.

**3.** Identify the key fields for the analysis:  
   a. Education: **ko_perus** (numerator) and **ko_ika18y** (denominator).  
   b. Income: **hr_pi_tul** (numerator) and **hr_tuy** (denominator).  

- *The numerator is the specific group you are counting (e.g., people with low education), and the
denominator is the total relevant population from which that group is drawn (e.g., all adults).*  
- ***ko_ika18y** and **hr_tuy** both contain the same values. However, for clarity you might want to use
the one with the same prefix as the one you are comparing it to.*

**Cleaning up the data**

*The raw data contains placeholder values (-1) where data has been suppressed for privacy. Using
these values in calculations would give incorrect results, so we must first filter them out.*

1. The data uses -1 to mark cells with suppressed data for privacy (where the population is
too small). We need to remove these cells from our analysis.

2. Use ***select by expression*** to select the features that have values in the two wanted fields
(ko_perus, hr_pi_tul) that are 0 or greater AND have people living in them (ko_ika18y,
hr_tuy):

   <img src="figures/cleaningdata.png" width="300">



4. Save the selected features as a new layer and name it informatively.


**Calculating the proportions**

*Raw counts (e.g., 50 people with low education) are misleading because they don't account for
the total population of an area. Calculating percentages standardizes the data, allowing fair
comparison between dense and sparse neighborhoods.*

1. In the newly created layer, use the Field Calculator to calculate of people with only basic
education.  
- **Output field name**: edu_per (or similar, max 10 chars for Shapefile)  
- **Output field type**: Decimal number (real)  
- **Expression**: ( "**ko_perus**" / "**ko_ika18y**" ) * 100

2. Repeat the process to calculate the proportion of people in the lowest income category.  
- **Output field name**: inc_per   
- **Expression**: ( "**hr_pi_tul**" / "**hr_tuy**" ) * 100  

**Reclassifying Data into Quartiles**

*To identify the "lowest" areas, we need a clear threshold. By splitting the data into four equal
groups (quartiles), we can systematically define the bottom 25% of areas for each variable,
providing an objective criterion for our analysis.*

1. We will reclassify our proportion fields into 4 classes (quartiles), where Class 1 is the lowest
and Class 4 is the highest.
  
2. First, find the quartile boundaries for the **edu_per** field.  
- Go to Layer Properties > Symbology.  
- Set it to Graduated, select the **edu_per** field, and set the Mode to Quantile (Equal
Count) with 4 classes.  

<img src="figures/equalcount.png" width="700">  

- Click Classify. The values in the "Values" column are your quartile boundaries.
Write them down.

<img src="figures/edu_per.png" width="300">  

3. Repeat the previous steps for the **inc_per** field and note down its quartile boundaries

4. Next, use the Field Calculatorr with a conditional statement to create a new reclassified field for education.
- **Output field name**: edu_class  
- **Output field type**: Whole number (integer)  
- **Expression:**

<img src="figures/edu_class.png" width="500"> 

a. The logic behind the numbering of the classes should be the same across all the
reclassifications. In this instance our logic can be as follows: class 1 is the lowest
(e.g. lowest share of inhabitants with no qualification after basic-level studies) and
class 4 is the highest quartile (e.g. highest share of inhabitants with no qualification
after basiclevel studies).

NOTE: Use a dot as a decimal separator instead of comma.  

b. Check the attribute table that the values are correct (click the new column field
name to order the values in order to see the highest and lowest values, you should
only have numbers 1, 2, 3 and 4).  

5. Repeat previous step to create a new field inc_class for the income proportions, using
the boundaries you found.  

**Finding and Mapping the Target Areas**

*The core of our analysis is to find areas that are disadvantaged in multiple ways. By selecting
cells that are in the worst quartile for both income AND education, we identify the areas facing
compounded socio-economic challenges.*


1. Our goal is to find cells that are in the lowest quartile for both education and income (i.e.,
where edu_class = 4 AND inc_class = 4).  
- Use Select by Expression.  
- Use the expression: " edu_class " = 4 AND " inc_class " = 4

2. Save the selected features as a new, final layer (e.g., low_edu_inc_areas).
