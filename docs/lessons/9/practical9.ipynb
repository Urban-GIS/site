## Prioritising urban green areas through patch-based indicators: Composite index

## OVERVIEW & PURPOSE

How can cities decide where to focus their nature conservation efforts? **Spatial Conservation Prioritization (SCP)** provides a systematic, data-driven answer. SCP is a branch of conservation biology that seeks to develop an optimal conservation solution by balancing various conservation objectives and constraints and turning the information into a map.

A simple SCP approach maps species distributions and optimizes their protection, but this alone may not capture urban biodiversity. Urban ecosystems are complex and dynamic, with ecological communities supporting key functions like pollination and decomposition that underpin ecosystem services. Therefore, conservation should focus on preserving these functioning communities. In GIS-based SCP, this can be addressed by mapping attributes such as species richness and biomass to represent biodiversity quality, allowing spatial prioritization to reflect ecosystem functioning rather than individual species alone.

These kinds of maps have been done for the Helsinki Metropolitan Area, for example, by Jalkanen et al. (2025) in their study "Identifying priority urban green areas for biodiversity conservation and equitable recreational accessibility using spatial prioritization" (Landscape and Urban Planning, 259, 105356 (2025). https://doi.org/10.1016/j.landurbplan.2025.105356)
See also Jalkanen's blog post for further reflection on the topic (in Finnish): [https://jalkanen.blogspot.com/2025/04/uusi-julkaisu-paakaupunkiseudun.html](https://jalkanen.blogspot.com/2025/04/uusi-julkaisu-paakaupunkiseudun.html).

**In summary, urban SCP uses GIS methods to identify the most valuable green spaces and habitat networks within cities, where land is limited and highly fragmented.** In addition to mapping direct measures of biodiversity, GIS methods can be used to:
- **Map urban habitat features** (e.g., parks, street trees, green roofs, water bodies)
- **Assess habitat quality** using indicators like vegetation structure or biodiversity potential
- **Analyse habitat connectivity**, showing how green areas are linked or isolated across the urban matrix
- **Identify pressures**, such as built-up areas, roads, and human disturbance

In this exercise, we will create a simple composite index based on urban green structure in Central Helsinki. A composite index integrates several variables or indicators into one score:

Green Structure Index = w<sub>1</sub>X<sub>1</sub> + w<sub>2</sub>X<sub>2</sub> + ... + w<sub>n</sub>X<sub>n</sub>

- X = indicators (e.g., patch size, connectivity)
- w = weights (importance of each factor)

The resulting map or dataset shows priority levels for green area development. 

We will create the index based on three variables: patch size, patch connectivity, and vegetation structure. Patch size and vegetation structure are positively related to the number of habitats and, thus, the number of species and biodiversity maintained by a green area, as well as the ecosystem services it provides. Larger patches are also less vulnerable to disturbance compared to small patches. Patch connectivity enables species movement and gene flow between the green areas, supports the functioning of ecological networks, and reduces the effects of isolation.

### Creating indicators

#### Indicator 1: Patch size
4. We will first calculate the size of each green area using the *Field calculator* tool in the Attribute Table.
    1. Create a new field in the Attribute Table, for example, area_ha.
    2. Use the expression *$area*, found under the Geometry drop-down menu. The unit will be square meters. We prefer hectares, so we will have to convert it by multiplying by 10,000.
        1. Use a decimal number as the output field type to get precise results.
        2. **Tip:** The output can be rounded using the round() function.

5. We will then reclassify the green areas by size.
    1. Create another new field, for example, size_score, and create an expression that ranks the green areas based on their size.

    > The principle is that the bigger the green area is, the higher it should be prioritized in urban planning, and thus the higher score it should be assigned.

    > Note that different ecosystems and ecosystem services require different-sized green areas. There is no one optimal size, rather, the size depends on the phenomena under study, and both, small and large green areas are important as part of a green network. For example, Sari & Bayraktar (2023) have studied the role of park size on ecosystem services in urban environment in an article "The role of park size on ecosystem services in urban environment: a review" (*Environ Monit Assess 195, 1072 (2023). https://doi.org/10.1007/s10661-023-11644-5.*)

    2. You may define the ranking yourself, or use the following example, based on Sari & Bayraktar's (2023) study:

     |Green area size (ha) | Description | Score |
     |------|------|------|
     | < 1 | Small, supporting green area | 1 |
     | 1–10 | Green area supporting human activity and urban biodiversity | 2 |
     | > 10 | Green area supporting stable ecosystems and strong ecosystem services | 3 |
    
        In this case, the expression would be:

        CASE
         WHEN "area_ha" >= 10 THEN 3
         WHEN "area_ha" >= 1 THEN 2
         ELSE 1
        END
        
#### Indicator 2: Patch connectivity

6. We will measure the connectivity of the green areas by counting the neighbouring green areas within a buffer distance from each green area.
    1. Create buffers using the *Buffer* (Vector Geometry) tool. You may define the buffer distance yourself. For example, the distance of 300 m corresponds to the mobility of many species in urban environments and represents a commonly used scale in urban ecological analyses.
    2. Then we will count the number of neighbouring green areas within each buffer using the *Join attributes by location (Summary)* (Vector general) tool.
        1. Join the *intersecting* features in the green area data by comparing to the buffer data.
        2. Select a field to summarize, for example, fid, and select *count* as the summaries to calculate.
        3. As a result, a new column will be generated in the output layer's Attribute Table.

7. We will then reclassify the green areas based on the number of neighbouring green areas.
    1. Create a new field in the Attribute Table, for example, connectivity_score, using the Field Calculator, and create an expression that ranks the green areas based on their connectivity.
   
    > The principle is that the more connected the green area is, the higher it should be prioritized in urban planning, and thus the higher score it should be assigned.
    
    2. You may define the ranking yourself, or use the following example:

    | Number of neighbouring green areas | Description | Score |
    |------|------|------|
    | < 5 | Weak connection | 1 |
    | 5–10 | Moderate connection | 2 |
    | > 10 | High connection | 3 |

    Form a corresponding expression.

#### Indicator 3: Vegetation structure

8. We will evaluate vegetation structure in the green areas using the original land cover data with vegetation height and the *Zonal Statistics* tool (Raster analysis).
    1. The algorithm calculates statistics of a raster layer for each feature of an overlapping polygon vector layer.
    2. We are going to calculate which vegetation height class covers the majority of the green areas. The vegetation height raster classes are:
   - 213 vegetation < 2 m
   - 231 vegetation 2–5 m
   - 232 vegetation 5–10 m
   - 233 vegetation 10–15 m
   - 234 vegetation 15–20 m
   - 235 vegetation > 20 m       
    3. Select the vegetation landcover data as the input layer, and the original landcover data as the raster layer.
    4. Name the *Output column prefix*, for example, vegetation_, and select Majority as *Statistics to calculate*.
    5. As a result, a new column will be generated in the output layer's Attribute Table.

9. We will then reclassify the green areas based on vegetation height, indicating the vegetation structure.
    1. Create a new field, for example, vegetation_score, using the Field Calculator, and create an expression that ranks the green areas based on the vegetation height.
    
    > The principle is that the higher the vegetation is in general, and thus the more complex the vegetation structure is, the higher it should be prioritized in urban planning, and thus the higher the score it should be assigned.

    2. You may define the ranking yourself, or use the following example:

    | Vegetation height (m) | Description | Score |
    |------|------|------|
    | < 5 | Low vegetation | 1 |
    | 5–15 | Moderately high vegetation | 2 |
    | > 15 | High vegetation | 3 |

    Form a corresponding expression. Note that this time the expression is based on raster classes, not numeric values.
                                                                                                                                                                                                                                                                                                                     
### Creating composite index

10. Before creating the composite index, we need all three indicator scores in the same vegetation data. If not, join the missing columns.
    1. Go to *Layer Properties* and open *Joins* tab.
    2. Add a new join. Select the layer that contains the score, and select *fid* as both the join and target field.
    3. We will only need the score field, which can be selected under the *Joined fields*. The column prefix can be edited under the *Custom field name prefix*.
    4. Save the vegetation layer into a new layer to save the joined columns permanently into the data.
    
11. Now we can combine the three indicators into an index.
    1. Create a new field, for example, green_index, using the Field Calculator, and create an expression to calculate the composite index.
    2. At this point, the index can be applied with weights. The patch size could, for example, be considered more important (0,5) than the patch connectivity (0,25) and vegetation structure (0,25) indicators. You may define the weights yourself.
        1. The expression for calculating the weighted sum for the suggested index is:

        green_area_index = ("size_score" * 0,5) + ("connectivity_score" * 0,25) + ("vegetation_score" * 0.25)

        2. Use a decimal number as the output field type to get precise results.

### Classifying the index

12. Finally, after calculating the composite index, we can reclassify the green areas according to their index values or prioritisation.
    1. Create a new field, for example, priority_class, using the Field Calculator, and create an expression to classify the index.
    2. The classification expression could look something like this:

    CASE
     WHEN "green_area_index" > 2 THEN 'High'
     WHEN "green_area_index" > 1 THEN 'Medium'
     ELSE 'Low'
    END
