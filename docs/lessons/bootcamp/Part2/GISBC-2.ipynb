*Combining spreadsheet data (Excel files etc.) with spatial data is a common operation in GIS.* For
instance, you might have a spreadsheet containing air quality measurements from measuring stations and
a point feature shapefile containing the location data of these measuring stations. You want to visualize
the measurements on a map and thus you have to somehow combine the data in GIS.

### III. Acquiring data for GIS

#### Open data, and where to find it

GIS can process many different kinds of data. Most common data types are vector data (points, lines and
polygons) and raster data (usually represented as pixels, such as aerial imagery). All kinds of tabular data
(Excel, .txt, .csv) can also be easily added to GIS with some “cleaning”.

You have already practiced adding some data to QGIS last week. However, that data was already found
and zipped for you in GIS-friendly forms, shapefiles (.shp). In this part, we will go through some of the
ways to acquire data for various GIS analyses.

**As QGIS is an open-source GIS software and we mostly use open data during this course, it’s good to  
know where you can find open spatial data to work (or have fun) with.**  

The biggest problem with open data is its dispersed nature: you can’t find all open data under one data  
portal. Numerous data providers have their own services to disseminate their data, and many services  
require you to register a user account. In the worst-case scenario, just to get a good collection of data  
you’ll have to create numerous accounts in one sitting, wait for your registration to be acknowledged and  
then wait for your data request to be handled. After you’ve received the data you need, it usually requires  
some further processing to make it ready for analysis. 

There are some data portals that have loads of data from different data providers under one roof. For
Finnish open spatial data, one of the best download services is **PaITuli**, and one the best viewing services
is **Paikkatietoikkuna.** There are also **Lounaispaikka** (South-Western Finland) and **Helsinki Region
Infoshare.** For general open data about Finland, there is **Avoindata.fi.**

Most of the sites and data providers we’ve mentioned have a web-browser-based download/viewing
service, but for some, there is another way. **There are services for viewing and downloading spatial data
directly in QGIS**, without having to use the web browser and continuously download zip-packages:

&nbsp;&nbsp;&nbsp;&nbsp;• **WFS (Web Feature Service):** provides **vector** data (such as shapefiles), which can be downloaded  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;and saved onto your computer.  

&nbsp;&nbsp;&nbsp;&nbsp;• **WCS (Web Coverage Service):** allows you to download **raster** data (satellite/aerial images etc.).  

&nbsp;&nbsp;&nbsp;&nbsp;• **WMS (Web Map Service):** provides raster data for viewing – but it will not allow you to download  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;the data.

### IV. Vector clipping

When working with GIS and large datasets, but with a small area of interest, it’s a good idea to extract
the area of interest from the large dataset for processing instead of working with the original massive
dataset. This can be done in various ways both with vector and raster data, but a commonly used method
is to use Clip to extract the area of interest.


For instance, you might have a vector dataset of all the buildings and land coverage of Finland, but all you
need is the data from one particular area. The area could be your study area, a municipality or something
else altogether. In order to extract the data from the area of interest, a mask layer is required. A shapefile of the municipality’s area and borders can be a mask file, or you can make one yourself if the area of
interest is a study area. Let’s try a basic clipping operation with the data we have at hand.

**13. Clip a layer of all roads of the metropolitan area into one that only includes the roads of Helsinki:**

&nbsp;&nbsp;&nbsp;&nbsp;**13.1** Add the following layers into the project: *Helsinki_Municipality.shp, PKS_roads.shp*. They are in the course data folder.<br>
&nbsp;&nbsp;&nbsp;&nbsp;**13.2** In Processing toolbox, type *Clip* into the search bar. Choose QGIS’s own algorithm under *Vector overlay tools*. Various different options<br>
&nbsp;&nbsp;&nbsp;&nbsp;appear here for clipping operations by different algorithm providers, some of them are for raster/grid data and some are for vector data.  
&nbsp;&nbsp;&nbsp;&nbsp;**13.3** After opening the algorithm, select *PKS_roads* as Input and *Helsinki_Municipality* as Clip layer. This means we’re creating a new shapefile  
&nbsp;&nbsp;&nbsp;&nbsp;that will only include the roads from *PKS_roads* within Helsinki’s borders and nothing else.<br>
&nbsp;&nbsp;&nbsp;&nbsp;**13.4** After running the algorithm, remember to save the temporary *“Clipped”* layer as a new shapefile with an informative name  
&nbsp;&nbsp;&nbsp;&nbsp;(*Right click > Export > Save features as*).

### V. Working with raster data

#### Getting started

In this part we will take a look at how to work with and visualize raster type data in QGIS, specifically the
digital elevation model (DEM) from the National Land Survey of Finland. Open spatial data are typically
distributed in tiles of manageable file sizes if the data is heavy, which is often the case for raster type data
(aerial images, DEMs, etc.). Tiling allows the user to choose and download only the required data, but it
is often necessary to combine the tiles into a single raster for further processing.

The DEM tiles covering the Helsinki area or any area in Finland can be downloaded from the Paituli spatial
download service (https://paituli.csc.fi/download.html). DEM tiles can be found in the exercise
folder (Helsinki_dem), if you are interested in examining them. Today, we will not dig too deep into
combining raster layers into a merged DEM raster: it has already been made and can be found in the
exercise folder.

However, it is important to notice that working with raster data is slower than with vector data, since the
data is much heavier, and processes take longer. Let’s try a simple raster analysis.

A common way to visualize a DEM is using a so called hillshade effect, which imitates the effect of sunlight
on terrain. In a hillshade image, the raster cell values are determined by the slope aspect and steepness,
the actual terrain height information is lost.

One way to produce a hillshade effect in QGIS is directly in the Layer Properties’ Symbology tab by
changing the Render type to Hillshade. This is quick and easy but does not actually create a new raster file
and does not allow for further processing. Let’s instead use a processing tool.

### EXTRA

QGIS’s functionality is quite extensive out-of-the-box, but it can be extended ever further with plugins.
For instance, integrating CAD functionality into QGIS can be done with plugins to some degree. The plugins
are written by QGIS users and community members; some are more official while others are not. Some
very popular plugins have been incorporated into QGIS by default.

In the main QGIS window/work view, there’s a drop-down menu Plugins in the top of the window, and
under that, there’s Manage and install plugins. Click it. A window opens up with a few tabs: All, Installed,
Not installed, Upgradeable, and Settings. The All tab contains all the plugins and doesn’t sort them by
whether they’re installed on your computer or not. The Upgradeable tab houses the plugins that you’ve
already installed but for which you have a newer version waiting to be downloaded. The Settings tab has
a few plugin-related settings, such as automatic update checks, experimental plugins and deprecated
plugins. Update checking should be ticked, and a preferred update check interval selected, this way you’ll
stay up to date with your plugins.

Experimental plugins are plugins in differing stages of development and shouldn’t be used unless you
know exactly what you’re doing. Deprecated plugins are old plugins for previous QGIS versions or plugins
that aren’t being updated anymore. You shouldn’t use deprecated plugins unless you know what you’re
doing. There’s also a section called Plugin repositories, which holds all the repositories QGIS checks for
downloadable plugins. By default, it houses the QGIS’s official plugin repository, but some 3rd party
repositories are out there as well. 3rd party repositories can be useful, but the official repository – being
by far the largest – will satisfy most QGIS users. Some companies, such as Boundless Spatial, have their
own repositories.

#### Browsing plugins

Go to All tab in the Plugins window and you’ll be faced with a large number of
different plugins. Some plugins are very simple and general, while others are very
complex and specific. For instance, there’s a plugin which enables the creation
of heatmaps, as well as a plugin custom-made for the Brazilian Army’s
cartographic department. Some of the names of the plugins are written in green
text, which means the plugins are “Trusted”. While there is a review process
before a plugin gets accepted into the official repository, it can’t catch every
problem or possible threat in bad code. Plugins marked with “Trusted” are
developed by well-known QGIS community members and the code is of good
quality. Plugins which are not marked with “Trusted” may not be untrustworthy,
but it’d be good to do some research on the plugin before installing it.

There’s a search bar you can use to find plugins close to your interests. Typing in CAD, hydro or whatever
you find interesting might yield a plugin for that particular purpose. There are some must-have plugins (at
least from a geographer’s perspective): OpenLayers, QuickMapServices, Heatmap, XyTools, Metatools,
NNJoin, Point sampling tool, Statist, Raster Interpolation and Terrain Analysis. You can search them using
the search bar and see what functionality they bring. Some of the plugins might be installed by default
but be disabled. Tick the box next to the name of the plugin to enable its functionality.

For architects, Dxf2shp converter might be a valuable plugin, as is anything that enables CAD-GIS
interoperability. Ecologists or physical geographers might find Crayfish’s hydrological raster tools
interesting. Archaeologists might find AGT and ArcheoCAD helpful. Lastly, here are some fun plugins
everyone can enjoy: Qgis2threejs is a fun plugin to experiment with javascript-based 3d functionality. You
can create a 3D model out of a DEM, add vector data into it and impress others. If you have access to a
3D printer and want to create something cool geo-geeky: DEMto3D creates a 3D printer friendly file of a
DEM raster, you could print a miniature version of the Grand Canyon as an ornament for your apartment.
Colorize it, buy a frame, and hang it on a wall?
