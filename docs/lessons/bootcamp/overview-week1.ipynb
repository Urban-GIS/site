## Lecture 1: Introduction to GIS

This lecture begins with an introduction to the course and the fundamentals of Geographic Information Systems (GIS). We explore what makes data geographic, how spatial and attribute data work together, and how real-world features are represented using vector and raster data. We also introduce maps, scale, generalization, and coordinate systems.

The lecture continues with **GIS Bootcamp 1**, a hands-on introduction to QGIS. You will become familiar with the QGIS interface, add and explore spatial data, apply symbology, perform basic calculations and spatial operations, and create a map layout.

### Learning objectives

By the end of this lecture and GIS Bootcamp 1, you should be able to:

* Explain what GIS and geographic data are.
* Recognize common spatial data types and representations.
* Describe some key applications of GIS.
* Navigate the QGIS interface and manage a QGIS project.
* Add, explore, visualize, and analyze basic spatial datasets.
* Create a simple map output in QGIS.
