### I. The basics of QGIS
#### Getting started with QGIS

**1.Download the data zip file containing the crash course [data](https://github.com/Urban-GIS/site/releases/download/course-data/data_bootcamp.zip)**

**2. Unzip the data to your computer’s hard drive or to an external hard drive/memory stick.**

*Wherever you save, make sure you have enough space left there: the package is about 2 GB uncompressed.*

**3. Open QGIS.**

When you open QGIS, the QGIS graphical user interface (GUI) will open (see Figure 1). 

<img src="figures/GUI.png" width="600">

*Figure 1. QGIS GUI and its main sections.*

**4. Change the default language in QGIS into English (highly recommended for this course):**
  
   **4.1** Navigate to the language settings: Settings menu > Options > General > User Interface Translation.

   **4.2** Tick the “Override system locale” box. Choose English from the dropdown menu. Close the window by pressing OK.

   **4.3** Restart QGIS for the change to take effect.

*As Qgis is developed by a community of volunteers, it is continuously being translated, and the GUI as well
as the user manuals are already available in several different languages.*

**5. Create a new project:**

<img src="figures/newproject.png" width="60"> **Creating and opening a project in QGIS.** To start a new project, you can

&nbsp;&nbsp;&nbsp;&nbsp;a. click on the *New* icon, which has a blank page (pictured above), or  
&nbsp;&nbsp;&nbsp;&nbsp;b. go to *Project > New*, or  
&nbsp;&nbsp;&nbsp;&nbsp;c. use *Ctrl + N** (or *Command + N*)

To open an already existing project, you would click on the folder-like Open icon (pictured above).

*The state of a working session in QGIS is called a ‘project’. Similarly, to a e.g. a ‘workspace’ in ArcGIS, a
project is considered the ensemble of layers, projections, table relations and other properties, such as
symbols and styles, of a specific session.*

**6. Save the project:**

<img src="figures/saving.png" width="30"> ***Saving in QGIS.*** To save your project, you can 

&nbsp;&nbsp;&nbsp;&nbsp;a. click on the *Save* or *Save As* icon (pictured above), or  
&nbsp;&nbsp;&nbsp;&nbsp;b. go to *Project > Save*, or  
&nbsp;&nbsp;&nbsp;&nbsp;c. use *Ctrl + S* (or **Command + S*)

*The file format of a project file is **.qgs.** Always remember to save your projects often to prevent work from
being lost in case of a crash. Note that project files do not contain geospatial data, they merely contain
information on where the program will find it. If you later move the location of the used data on your
computer, QGIS might not find the data when you re-open the project.*


**7. Change the coordinate system (CRS) of your project to EPSG:3067**

<img src="figures/EPSG.png" width="80"> ***Changing the coordinate system of a project in QGIS.*** You can see the current coordinate
system of the project in the status bar in the lower right corner. To change CRS, you can

&nbsp;&nbsp;&nbsp;&nbsp;a. click on the sign (pictured above), or  
&nbsp;&nbsp;&nbsp;&nbsp;b. go to project > properties > CRS, 
and then then select a new coordinate system from the list in the project properties window that
pops open.

The default coordinate system in QGIS is automatically set to EPSG:4326 (WGS 84). You can change the
default to EPSG:3067 if you wish by going to Settings > Options > CRS. The data used in this course will
mostly be in EPSG:3067 (ETRS-TM35FIN), which is the standard for nationwide data in Finland.

*QGIS supports ‘on the fly’ (OTF) coordinate system transformation for both vector and raster layers. This
means that the program will automatically draw the layers to match the coordinate system defined for
the map canvas, if a transformation formula is available.*

#### Field calculator, select by expression and spatial join

Next, we shall move the focus from visualization to the actual data.

***17. Examine what data is included in the project and what information is stored in the layers by right

**18. In the attribute table, toggle Editing mode** <img src="figures/toggleediting.png" width="30">  **and then click on the Field Calculator button** <img src="figures/fieldcalc.png" width="30"> , which is in the Attribute table toolbar in the top part of the window.

*The Field Calculator is a tool that allows you to create new attribute fields in an attribute table or updating existing ones by applying calculations or expressions.*

**19. Write an expression that calculates the area of each small area of Helsinki in square kilometers:**

&nbsp;&nbsp;&nbsp;&nbsp;**19.1** Type Area_km2 into Output field. Set the Output field type to Decimal number (real) and
Precision to 2.

&nbsp;&nbsp;&nbsp;&nbsp;**19.2** On the right side of the Expression window is a list of drop-down menus. Click the Geometry
drop-down menu open.

&nbsp;&nbsp;&nbsp;&nbsp;**19.3** Select `$area` expression by double-clicking it. Alternatively, you can type `$area` in the blank
Expression window.

&nbsp;&nbsp;&nbsp;&nbsp;**19.4** If you use only `$area`, the result will be in square meters. To convert it to square kilometers, divide by `1 000 000`. Click the division symbol `/` above the Expression window or type it manually, then enter `1000000` and click **OK**.

&nbsp;&nbsp;&nbsp;&nbsp;**19.5** Click *Save Edits* button <img src="figures/saveedits.png" width="35"> and disable *Editing mode.* <img src="figures/toggleediting.png" width="35">
