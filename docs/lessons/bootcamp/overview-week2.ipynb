## Lecture 2: Spatial Data, Open Data, and Coordinate Systems

This lecture introduces common sources and formats of spatial data, including open-data portals and web services such as WMS, WFS, and WCS. We then examine coordinate reference systems, map projections, and the distortions that projections can introduce.

The lecture continues with **GIS Bootcamp 2**, where you will practice acquiring open spatial data and working with vector and raster datasets in QGIS. You will also digitize features, clip datasets to a study area, perform basic raster analysis, and visualize the results.

### Learning objectives

By the end of this lecture and GIS Bootcamp 2, you should be able to:

* Identify common sources and formats of spatial data.
* Explain the differences between WMS, WFS, and WCS services.
* Explain why coordinate reference systems and map projections matter.
* Find and add open spatial data in QGIS.
* Create and edit vector features and their attributes.
* Clip vector and raster data to a selected study area.
* Perform basic raster analysis and visualize the results in QGIS.
