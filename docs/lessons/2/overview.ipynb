## Week 4 overview

### Learning objectives

**Lecture**
- Understand visual variables (shape, size, color hue/brightness/saturation, texture, orientation) and visual isolation in map symbol design
- Apply visual hierarchy principles to guide a map reader's eye and structure
- Follow the geographic inquiry process: evaluate the phenomenon → define purpose/audience → collect & process data → design & construct the map → evaluate effectiveness
- Recognize layout best practices (grid/alignment, avoiding filled corners, contrast via hue/saturation/value, centering key information)

**Practical**
- Reinforce independent QGIS workflow (minimal step-by-step guidance)
- Use Zonal Statistics to extract raster (Corine land cover) summary values into a vector layer's attribute table
- Refresh Field Calculator skills: computing area ($area), unit conversion (m² ↔ km²), and derived fields
- Join attributes by location (summary) to bring population data (HSY grid, "ASUKKAITA") into small-area polygons
- Calculate a population-pressure-on-green-space index (inhabitants per km² of green area)
- Build a styled, print-layout map communicating the result
- Write a short analysis report answering interpretation and reflective questions
