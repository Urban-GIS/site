### OVERVIEW & PURPOSE

In this exercise, your task is to find out the amount of population pressure towards green areas in the 
city of Helsinki. To achieve this, you will learn to utilize zonal statistics with raster data and refresh your 
memory on how to use the field calculator.

*Access to green space within cities has been found to benefit many aspects of health and wellbeing, 
enabling local residents to cope better with the stresses of living in large urban areas. To predict the 
consequences of rapid urbanisation for human wellbeing requires information on how green space 
provision will change as cities grow. One way to help achieve this is to study green space provision across 
a range of present-day cities.* European Commission, Fuller & Gaston 2009.

This is the second actual exercise of the course and as such, the difficulty level has been raised ever so 
slightly. The slightly risen difficulty materializes itself as less-detailed instructions to reinforce your 
learning. In this way you’ll have to recall, think and/or discover how different operations are done in QGIS. 

If you feel stuck, this is normal!
This practical is designed to help you become a more independent GIS user. Before asking for help, try 
these steps:
1. **Revisit the Crash Course Materials:** The answer is often there.
2. **Use the QGIS Help Menu:** Search for the tool name.
3. **Search Online:** use online help by searching the tool name together with QGIS. (e.g."join attributes by location QGIS.)  
4. **Ask on Discord:** Describe the problem, what you've tried, and include a screenshot.

### EXERCISE PHASES

**Green areas**
1. Download and open the exercise 2 data in QGIS from the course Moodle page (and explore its 
attribute tables to familiarize yourself with what sorts of data you actually have). Import the shapefiles 
and the TIF file. Remember to use the correct import method based on data types.
2. Use the field calculator to calculate the area (m2) of each small area of Helsinki <img src="figures/fieldcalc.png" width="35">. Create a new attribute field for area values and use the $area expression in the field calculator.
3. Use **Zonal Statistics** tool to calculate how many pixels (count) depicting green areas (Corine data) are 
 located within each small area of Helsinki 
- Use the Helsinki small area polygons as the input layer and the CORINE green area raster as the raster layer.
- Select *Count* in the Statistics to calculate parameter and click run.
- Save the layer as a new permanent layer and add it to your project.
4.  Use the field calculator to calculate the area (m2) of the green areas using the number of green area pixels. 
- Go to layer properties and check the pixel sizes.
- Use the layer that you made with Zonal Statistics and create new field and name it informatively.
- Then calculate the green area by multiplying the pixel count by the area represented by one raster pixel.
<img src="figures/greenpix.png" width="500">  

5. Use the field calculator to calculate the proportion of green areas to the actual area of the small areas.  
- Create a new field for the green area percentage.      
- Divide the green area by the total area of the small area and multiply by 100.     


**Population data** 

6. Use **join attributes by location (summary)** to join the population data into the small areas of 
Helsinki layer. (Hint: you need the sum of “ASUKKAITA” field)

- Use the layer you updated in the previous section containing the green areas.
  - Remember to save the changes while you are in the Attribute table and toggle editing 
off *(the pencil symbol)*.
- Select the Population layer as the Join layer and select “ASUKKAITA” field to summarize.
- Select *Sum* and click *Run*.
- Save the layer as a new permanent layer and add it to your project.


7.  Using the field calculator, calculate how many inhabitants there are for each square kilometer of 
green area within the small areas of Helsinki. This field will indicate the population pressure towards green areas:

- Create a new field and check the conversion rules from square meters to square kilometers.
- Calculate the population per square kilometer of green area.


**Visualization**

8. Visualize the population per green area on map using the **print layout**. You can also visualize the 
proportions of green areas onto a map. Having both maps can help you when writing the report.
